# Explore fertility results (exploratory)

Load the latest experiment folder and inspect CSVs. Not authoritative for the paper.

```python
from pathlib import Path

import pandas as pd

exp_root = Path("experiments")
runs = sorted(exp_root.glob("*_*"), key=lambda p: p.stat().st_mtime, reverse=True)
if not runs:
    print("No experiment runs yet — run scripts/02_run_tokenizer_audit.py first")
else:
    latest = runs[0]
    df = pd.read_csv(latest / "results" / "fertility_by_register.csv")
    display(df)
```
