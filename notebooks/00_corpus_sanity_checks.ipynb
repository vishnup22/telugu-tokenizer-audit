# Corpus sanity checks (exploratory)

This notebook is **not** the source of truth for paper numbers.

```python
from pathlib import Path

from telugu_audit.corpus.loaders import REGISTERS

corpus_dir = Path("data/processed")
if not (corpus_dir / "native_formal.txt").exists():
    corpus_dir = Path("tests/fixtures/toy_corpus/_processed")

for register in REGISTERS:
    path = corpus_dir / f"{register}.txt"
    if path.exists():
        lines = path.read_text(encoding="utf-8").strip().splitlines()
        lengths = [len(line) for line in lines]
        mean_len = sum(lengths) / max(len(lengths), 1)
        print(register, "n=", len(lines), "mean_len=", mean_len)
```
